# Accuracy

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

# Nilai K
k_values = data["Tanpa Normalisasi"]["K"]

# Membuat figure
plt.figure(figsize=(10, 6))

# Plot masing-masing metode normalisasi
plt.plot(
    k_values,
    data["Tanpa Normalisasi"]["Accuracy"],
    marker="o",
    linewidth=2,
    markersize=7,
    label="Tanpa Normalisasi"
)

plt.plot(
    k_values,
    data["Min-Max Scaling"]["Accuracy"],
    marker="s",
    linewidth=2,
    markersize=7,
    label="Min-Max Scaling"
)

plt.plot(
    k_values,
    data["Z-Score Scaling"]["Accuracy"],
    marker="^",
    linewidth=2,
    markersize=7,
    label="Z-Score Scaling"
)

plt.plot(
    k_values,
    data["Decimal Scaling"]["Accuracy"],
    marker="D",
    linewidth=2,
    markersize=7,
    label="Decimal Scaling"
)

# Judul dan label
plt.title(
    "Accuracy",
    fontsize=14,
    fontweight="bold",
    pad=15
)

plt.xlabel("Nilai K", fontsize=12)
plt.ylabel("Accuracy", fontsize=12)

# Menampilkan semua nilai K pada sumbu X
plt.xticks(k_values)

# Format sumbu Y menjadi persen
plt.gca().yaxis.set_major_formatter(PercentFormatter(1.0))

# Grid
plt.grid(
    axis="both",
    linestyle="--",
    alpha=0.4
)

# Legend
plt.legend(
    title="Teknik Normalisasi",
    fontsize=10
)

# Layout
plt.tight_layout()

# Tampilkan grafik
plt.show()
```
